```python
import pandas as pd

df = pd.read_csv("E:\Python_Projects\Employee Data Analysis System\employee_data_analysis.csv")
print(df)
```

```python
print(df.shape)

print(df.columns)
```

```python
print(df.dtypes)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Salary"] = df["Salary"].fillna(df["Salary"].mean())

df["Experience"] = df["Experience"].fillna(df['Experience'].mean())

df["City"] = df["City"].fillna(df['City'].mode()[0])

print(df)
```

```python
print(df.isnull().sum())
```

```python
print(df['Salary'].max())
print(df['Salary'].min())
print(df['Salary'].mean())
```

```python
# How many employees are there?
print(df.shape)
```

```python
# How many employees are in each department?
print(df.groupby('Department').count())
```

```python
print(df.groupby('Department').size())
```

```python
# Which department has the highest average salary?
print(df.groupby('Department')['Salary'].mean())
```

```python
# Show all employees earning more than ₹60,000.
print(df[df['Salary']>60000])
```

```python
# Show employees whose age is greater than 30.

print(df[df['Age']>30])
```

```python
# Find the Top 5 highest-paid employees.
print(df.sort_values(by='Salary',ascending=False).head(5))
```

```python
# Show employees with more than 5 years of experience.

print(df[df['Experience']>5])
```

```python
# Show employees who belong to the IT department.
print(df[df['Department']== 'IT'])
```

```python
# Show employees from Pune.
print(df[df['City']=='Pune'])
```

```python
# Find the average age of employees in each department.
avg_age = df.groupby('Department')['Age'].mean()
print(avg_age)
```

```python
# Which city has the highest number of employees?
highest_emp = df.groupby('City').size()
print(highest_emp)
```

```python
# Find the employee with the highest salary.
print(df.sort_values(by='Salary',ascending=False).head(5))
```

```python
# Which department has the most employees?
import matplotlib.pyplot as plt

dept_count = df.groupby('Department').size()
plt.title("Department Wise Employee")
dept_count.plot(kind = 'bar')
plt.show()
```

```python
# Which department pays the highest average salary?
dept_avg = df.groupby('Department')['Salary'].mean()
dept_avg.plot(kind='bar')
plt.title("Department Wise Average Salary")
plt.show()
```

```python
# How are salaries distributed?

df['Salary'].plot(kind='hist',bins=5)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of employees")
plt.show()
```

```python
# Which city has the most employees?
count_city = df['City'].value_counts()
count_city.plot(kind='bar')
plt.title("Employee By City")
plt.xlabel("City")
plt.ylabel("Number of employees")
plt.show()
```

```python
df.to_csv("clean_employee_data.csv", index=False)
```

```python
 # Top 5 Highest Paid Employees
top5 = df.sort_values(by="Salary", ascending=False).head(5)
print(top5)

top5.plot(
    x="Name",
    y="Salary",
    kind="bar",
    legend=False
)
plt.title("Top 5 Highest Paid Employees")
plt.xlabel("Employee Name")
plt.ylabel("Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
